```python
import matplotlib.pyplot as pit
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]
##create the line plot
pit.plot(x,y)
pit.xlabel("X axis")
pit.ylabel('Y axis')
pit.title("Basic Line Plot")
pit.show()
```

```python
pit.plot(x,y,color='red',linestyle='--',marker='o',linewidth=2,markersize=9)
pit.grid(True)
```

```python
x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,2,3,4,5]

pit.figure(figsize=(10,8))

pit.subplot(2,2,1)
pit.plot(x,y1,color='green',marker='o',markersize=5)
pit.title('Plot 1')

pit.subplot(2,2,2)
pit.plot(y1,x,color='yellow',marker='o',markersize=5)
pit.title('Plot 2')

pit.subplot(2,2,3)
pit.plot(y1,y2,color='blue',marker='o',markersize=5)
pit.title('Plot 3')

pit.subplot(2,2,4)
pit.plot(y2,y1,color='red',marker='o',markersize=5)
pit.title('Plot 4')
```

```python
catogeries = ['A','B','C','D','E']
values = [5,7,3,8,6]
## create a bar plot
pit.bar(catogeries,values,color='purple')
pit.xlabel('catogeries')
pit.ylabel('values')
pit.title('Bar Diagram')
```

```python
x=[1,2,3,4,5]
y=[2,3,4,5,6]
pit.scatter(x,y,color='Blue',marker='x')
```

```python
lebal = ['A','B','C','D']
size = [30,20,40,10]
colors = ['gold','yellowgreen','lightcoral','lightskyblue']
explode = (2,0,0,0) ##move out the first slice
pit.pie(size,explode=explode,labels=lebal,colors=colors,autopct='%1.1f%%',shadow=True)
```
