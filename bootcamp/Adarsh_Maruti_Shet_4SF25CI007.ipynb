30. Write a program that calculates the sum of the digits of a number input by the user using a while loop.



```python
num = int(input("Enter a number: "))
sum_digits = 0
while num > 0:
    digit = num % 10
    sum_digits += digit
    num = num // 10
print("Sum of digits =", sum_digits)
```

31. Write a program that checks if a number input by the user is a prime number using a for loop.



```python
num = int(input("Enter a number: "))
if num < 2:
    print(num, "is not a prime number.")
else:
    for i in range(2, num):
        if num % i == 0:
            print(num, "is not a prime number.")
            break
    else:
        print(num, "is a prime number.")
```

32. Write a program that prints the first n Fibonacci numbers, where n is input by the user.





```python
n = int(input("Enter the number of terms: "))
a = 0
b = 1
print("Fibonacci sequence:")
for i in range(n):
    print(a, end=" ")
    c = a + b
    a = b
    b = c
```
